```python
#17.Dicionário de Contagem de Palavras
#Conte a frequência de cada palavra em uma string fornecida pelo usuário.

string = input("Digite uma frase: ")
frase = string.split() #Utilizei o split para remover os espaços e deixar como uma lista

contador = {}
# Nesse loop ele vai analisar cada palavra da frase e vai verificar no dicionário. Se tiver adiciona mais 1, se não mantém 1;
for palavra in frase:
    if palavra in contador:
        contador[palavra] += 1;
    else:
        contador[palavra] = 1;

print(contador)
```

```python
#16.Número Primo
# Crie um programa que verifique se um número fornecido é primo.
# Numero primo so poder ser divida por 1 ou por ele mesmo
numero = int(input("Digite um número: "))
primo = True #Variavel de controle


if numero <= 1: # Se o número for menor ou igual a 1 sabemos que não é numero primo
    primo = False
else:
    for x in range(2,numero): #Um loop para verificar o resto de todas as divisiões até por ele mesmo.
        if numero % x == 0: # Se alguma opreação do resto for igual a zero então ele não é numero primo
            primo = False
            break

if primo == True:
    print(f"O número {numero} é primo")
else:
    print(f"O número {numero} não é primo")
```

```python
#19.Jogo de Adivinhação
# Implemente um jogo onde o usuário deve adivinhar um número gerado aleatoriamente pelo computador.
import random

numero = int(random.randint(1,10)) # Para sorte um número é utilizado uma biblioteca randorw e a função randint(com o parametro)

tentativas = 3 #Variavel de Controle

for x in range(tentativas): # Um loop para tentativa, cada vez que ele erra diminui a tentativa, se errar tudo exibe uma saida mostrando o numero escolhido. Caso acerte exibe uma saída e acaba o jogo
    jogada = int(input("Digite o número: "))
    if jogada != numero:
        print("Voce errou. Tente novamente")
        tentativas -= 1
    else:
        print("Parabéns, acertou em cheio")
        break

if tentativas == 0:
    print(f"Infelizmente acabou as suas tentativas. Os números escolhido era: {numero}")
```

```python
#21.Busca Binária
# Implemente a busca binária para encontrar um elemento em uma lista ordenada.
# Lista ordenada uma lista em ordem
# Busca binária seria uma estrategia mais inteligente para procurar o numero escolhido()

lista = [3,5,7,6,5,31,6,8,6,7]
lista.sort() # Deixar a listar ordenada

busca = int(input("Qual número quer encontrar: "))
encontrado = False
inicio = 0
fim = len(lista)-1 # passsar o parametro dos indices da lista

while inicio <= fim:
    meio = (inicio + fim)//2 # // para captar o valor inteiro da divisão

    if busca == lista[meio]:
        print(f"Numero encontrado na posição {meio}")
        print(lista)
        encontrado = True
        break
    elif busca < lista[meio]: # tem que chamar novamente a lista e colocar o parametro do meio
        fim = meio - 1 #Direciona para esquerda
    else:
        inicio = meio + 1 # Direciona para direita
```

```python
#22.Criptografia de Cifra de César
#Crie uma função para criptografar uma mensagem usando a cifra de César.
# Cifra de cesar e criptografia simples usada para transformar uma mensagem em outra mensagem "embaralhada"


def Cripto_Crifa(mensagem, chave):
    alfabeto = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"
    resultado = ''

    for letra in mensagem.upper(): # lower para pegar a mensagem e deixar todas as letras minusculas
        if letra in alfabeto:
            posicao = alfabeto.index(letra) #Localização do index da letra selecionado dentro do alfabeto
            nova_posicao = (posicao+chave) % 26 # criando uma nova posição com o deslocamento
            resultado += alfabeto[nova_posicao] # Adicionando a nova letra no resultado final
        else:
            resultado += letra

    return resultado

mensagem = input("Digite uma mensagem: ")
chave = int(input("Quantidade do deslocamento: "))
print(f"Mensagem Criptografada:", Cripto_Crifa(mensagem,chave))
```

```python
#23.Descriptografar Cifra de César
# Crie uma função para descriptografar uma mensagem cifrada com a cifra de César.

def Cript_Cesar(mensagem,chave):
    alfabeto = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"
    resultado = ""

    for letra in mensagem.upper():
        if letra in alfabeto:
            posicao = alfabeto.index(letra)
            nova_posicao = (posicao-chave) % 26
            resultado += alfabeto[nova_posicao]
        else:
            resultado += letra

    return resultado

mensagem = input("Digite a Criptografia: ")
chave = int(input("A quantidade de deslocamento selecionado:"))

print(f"A mensagem descriptografada é",Cript_Cesar(mensagem,chave));
```

```python
#25.Classificação de Números
# Classifique uma lista de números em três categorias: positivos, negativos e zeros.

lista = [5, -1, 8, 0, -7, -4, 3, 1, 21, -5 , -3]
positivo = []
negativo = []
zero = []

for numero in lista:
    if (numero < 0):
        negativo.append(numero)
    elif (numero > 0):
        positivo.append(numero)
    else:
        zero.append(numero)

print(f"Positivos: {positivo} \nNegativos: {negativo} \nZeros: {zero}")
```
