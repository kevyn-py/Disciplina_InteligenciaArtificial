```python
#7.Número Ímpar e Par
numero = int(input("Digite um número: "))

if numero % 2 == 0:
  print(f"O número {numero} é par! ")
else:
  print(f"O número {numero} é impar")
```

```python
#8.Verificação de Palíndromo
#Palindromo siginifica palavras que podem ler de trás para frente e não muda a palavra
#Em python para inverter a variável é usando [::-1]
palavra = input("Digite uma palavra: ")
intertida = palavra [::-1]
if palavra == intertida:
  print(f"A palavra {palavra} é Palindromo")
else:
  print(f"A palavra {palavra} não é Palindromo")
```

```python
#9.Fatorial de um número
# O fatorial tem que começar com 1 por conta que todo numero vezes 1 é ele mesmo. Com isso consegue calcular o Fatorial de um número
numero = int(input("Digite um número: "))
fatorial = 1

for x in range(1,numero+1):
  fatorial = fatorial * x

print(fatorial)
```

```python
#10.Sequência de Fibonacci
#Fibonacci é uma sucessão infinita de números em que cada termo depois dos dois primeiros é a soma dos dois anteriores
zero = 0
um = 1

for x in range(10):
  print(zero)

  proximo = zero+um
  zero = um
  um = proximo
```

```python
#12.Contagem de Caracteres
# O set serve para remover as palavras repetidas e o count para fazer a contagem de cada letra especifica na palavra
palavra = input("Digite uma palavra: ")

for letra in set(palavra):
  print(f"{letra} = {palavra.count(letra)}")
```

```python
#13.Ordenação de Lista
#Para ordenar uma lista utiliza o .sort
lista = [4,11,8,45,2,1,5,7,9,6]
lista.sort()

print(lista)
```

```python
#14.Soma dos elementos de uma lista
#Precisa criar uma variavel 0 para somar todos os numeros de uma lista e nao utiliza range pois ele vai percorrer a lista
lista = [2,4,5,6,7,8,9]
soma = 0

for x in lista:
  soma += x

print(soma )
```

```python
#15.Média de uma lista

lista = [2,6,7,9,3,4]
soma = 0
quantidade = len(lista)

for x in lista:
  soma += x

print(soma/quantidade)
```
