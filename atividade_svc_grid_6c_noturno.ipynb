```python
import numpy as np
import pandas as pd
from sklearn import datasets
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score
```

```python
# Pandas faz a leitura do csv e transforma em um dataframe
arquivo = "Crop_recommendation.csv"
df = pd.read_csv(arquivo);
```

```python
# Pega as features que vao ser utilizadas e coloca em x, e define a coluna "label" como a que vamos prever em y
x = df[
    [
        "Nitrogen",
        "phosphorus",
        "potassium",
        "temperature",
        "humidity",
        "ph",
        "rainfall"
    ]
]

y = df["label"]
```

```python
# Divide os dados em conjuntos de treino e de teste (80% treino, 20% teste)
x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size = 0.2,
    random_state = 42,
    stratify = y    # Preserva a distribuição das classes, padronizando os treinos e testes
)
```

```python
# *********** NAO SEI SE PRECISA DEIXAR *************
print(x_train.shape)
print(x_test.shape)

print(y_train.shape)
print(y_test.shape)
```

```python
# Cria o model SVC e faz o treino dele, com os dados de treino
modelo = SVC()
modelo.fit(x_train, y_train)
```

```python
# Faz as previsões baseado nos dados de teste x
y_pred = modelo.predict(x_test)
```

```python
# Compara e calcula a acuracia da resposta do modelo
accuracy = accuracy_score(y_test, y_pred)

print("Acuracia: ", accuracy)
```

```python
# Define os parametros que vamos testar
param_grid = {
    "C": [0.1, 1, 10],            # O quanto o modelo é rigoroso com penalização de erros
    "kernel": ["linear", "rbf"],  # Qual é a separação que será realizada
    "gamma": ["scale", "auto"]    # ************ NAO ENTENDI ESSE CARA ****************
}
```

```python
# cria o grid para testar as configurações do SVC e achar as melhores combinações
grid = GridSearchCV(
    SVC(),
    param_grid,
    cv = 5,     # Divide os dados em 5 partes
    n_jobs=-1,  # Usa todos os nucleos disponiveis, estava travando na hora da execução sem ele
    verbose=1   # Mostra as informações da execução
)
```

```python
# o grid rece os treinos e começa a testar as combinações
grid.fit(x_train, y_train)
```

```python
# printa os resultados qual foi a melhor combinação de parametro e o seu resultado
print(grid.best_params_)
print(grid.best_score_)
```

```python
# O grid utiliza a melhor combinação para fazer a predição nos dados de teste
y_pred_grid = grid.predict(x_test)

accuracy_grid = accuracy_score(y_test, y_pred_grid)

print("Acurácia do GridSearch:", accuracy_grid)
```
