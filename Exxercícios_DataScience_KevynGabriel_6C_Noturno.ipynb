```python
#19. Crie dois arrays de números inteiros aleatórios de tamanho 20 e calcule o ângulo entre eles (produto escalar normalizado).
import numpy as np

array_1 = np.random.randint(0,10,20)
array_2 = np.random.randint(0,10,20)

produto_escalar = np.dot(array_1, array_2)

comprimento_vetor_1 = np.linalg.norm(array_1)
comprimento_vetor_2 = np.linalg.norm(array_2)

cosseno_angulo = produto_escalar / (comprimento_vetor_1 * comprimento_vetor_2)

angulo = np.arccos(cosseno_angulo)
angulo_graus = np.degrees(angulo)

print(f"Array 1 : {array_1}")
print(f"Array 2 : {array_2}")
print(f"Produto escalado: {produto_escalar}")
print(f"Ângulo : {angulo_graus} graus")
```

```python
#12.Verifique quais colunas de um DataFrame contêm valores duplicados.
import pandas as pd

dados = {'Produto': ['Mesa','Cadeira', 'Teclado', 'Mouse','Monitor'],
         'Preco': [250,180,250,45,250],
          'Quantidade': [250,3,1,5,250] }

df = pd.DataFrame(dados)

for coluna in df.columns:
    if (df[coluna].duplicated().any()):
        print(f"A coluna '{coluna}' possui valores duplicados!")

print()
print(df)
```

```python
#13.Crie uma tabela dinâmica a partir do DataFrame, calculando a soma de Quantidade por Categoria.
import pandas as pd

dados = {'Produto': ['Mesa','Cadeira', 'Teclado', 'Mouse','Monitor'],
         'Categoria': ['Imovel','Imovel','Periferico','Periferico','Periferico'],
          'Quantidade': [2,3,1,5,5] }

df = pd.DataFrame(dados)
#df.pivot_table -> Criação da Tabela Dinâmica(values= a coluna que será somada, index= coluna que será as linhas, agg func= função de agragação)
tabela_dinamica = df.pivot_table(values='Quantidade',index='Categoria',aggfunc='sum')

print(tabela_dinamica)
```

```python
#14.Mescle dois DataFrames baseados em uma coluna comum chamada ID utilizando um join.
import pandas as pd

produto = pd.DataFrame({
    'ID': [1,2,3],
    'Produto' : ['Lapís','Apontador','Tesoura']
    })

quantidade = pd.DataFrame({
    'ID' : [1,2,3],
    'Quantidade' : [23,12,21]
})
# DataFrame Produto juntando com Dataframe Quantidade, usando o ID para saber qual linha corresponde a qual.
df_mesclado = produto.join(quantidade.set_index("ID"), on   = "ID")

print(df_mesclado)
```

```python
#15. Crie um DataFrame com dados faltantes e use a interpolção linear para preenchê-los.
#interpolação linear para estimar um valor desconhecido que está localizado entre dois pontos conhecidos.
import pandas as pd
import numpy as np

dados = {'Produto': ['Mesa','Cadeira', 'Teclado', 'Mouse','Monitor'],
         'Preco': [100,np.nan,320,np.nan,451],
          'Quantidade': [2,3,1,5,5] }

df = pd.DataFrame(dados)

#utilizando o interpolate() para calcular a interpolação e preenche-los
df['Preco'] = df['Preco'].interpolate()

print(df)
```

```python
#16.Crie um gráfico de barras a partir de um DataFrame usando a biblioteca Pandas.
import pandas as pd
import matplotlib.pyplot as plt
dados = {'Produto': ['Mesa','Cadeira', 'Teclado', 'Mouse','Monitor'],
          'Quantidade': [7,3,1,5,9] }

df = pd.DataFrame(dados)

#Criação do gráfico de barras a partir do DataFrame
df.plot(
    x = "Produto",
    y = "Quantidade",
    kind = "bar"
)
```

```python
#17.Adicione uma coluna ao DataFrame que calcule o imposto (5%) sobre o valor da coluna Total.
import pandas as pd

dados = {'Produto': ['Mesa','Cadeira', 'Teclado', 'Mouse','Monitor'],
         'Preco' : [25,30,21,45,23],
         'Quantidade': [1,3,1,5,1],
         'Total': [25,90,21,225,23], }

df = pd.DataFrame(dados)

#Adicionando uma coluna e calculandoi o imposto
df['Imposto'] = df['Total'] * 0.05

print(df)
```

```python
#18.Exiba apenas as linhas de um DataFrame onde o nome do produto começa com a letra ”B”.
import pandas as pd


dados = {'Produto': ['Mesa','Borracha', 'Teclado', 'Bolsa','Monitor'],
         'Preco' : [25,30,21,45,23],
         'Quantidade': [1,3,1,5,1],
         'Total': [25,90,21,225,23], }

df = pd.DataFrame(dados)

#Filtrando apenas produtos que começa com a Letra 'B'
df_filtrado = df[df['Produto'].str.startswith("B")] # str.startswitch(condição) verifica cada linha da coluna

print(df_filtrado)
```

```python
#19.Crie um DataFrame a partir de um dicionário de listas e calcule a média de cada coluna numérica
import pandas as pd
dados = {'Produto': ['Mesa','Borracha', 'Teclado', 'Bolsa','Monitor'],
         'Preco' : [25,30,21,45,23],
         'Quantidade': [1,3,1,5,1],
         'Total': [25,90,21,225,23], }

df = pd.DataFrame(dados)

media = df.mean(numeric_only = True) #numeric_only = True -> Calcule apenas as colunas númericas de um DataFrame

print(media)
```

```python
#20. Agrupe os dados por uma coluna categórica e calcule a mediana de uma coluna numérica.
import pandas as pd
dados = {'Produto': ['Mesa','Cadeira', 'Teclado', 'Mouse','Monitor'],
         'Categoria': ['Imovel','Imovel','Periferico','Periferico','Periferico'],
          'Quantidade': [2,3,1,5,5] }

df = pd.DataFrame(dados)

mediana = df.groupby('Categoria')['Quantidade'].median()
print(mediana)
```
