```python
#3. Crie um array de zeros com comprimento 15 e substitua os valores da 5ª a 10ª posição por 1.
import numpy as np

array_zero = np.zeros(15) #np.zeros(comprimento) - Uma array de zeros
array_zero[4:10] = 1 # para substituir o valor 1 chamando a variavel[coloca a posição que deseja alterar]

print(array_zero)
```

```python
#4. Gere um array de 100 números inteiros aleatórios e encontre a mediana e a moda dos valores.
import numpy as np

array = np.random.randint(0,100,100)
mediana = np.median(array) #np.median para encontrar a mediana da array

#Para encontrar a moda de uma array
valores, contagens = np.unique(array, return_counts=True)
moda = valores[np.argmax(contagens)]
print(f"Array: \n{array}")
print(f"Mediana: {mediana}")
print(f"Moda dos valores: {moda}")
```

```python
#6. Gere uma matriz 5x5 de números aleatórios entre 0 e 1 e encontre a soma de cada coluna
import numpy as np

array = np.random.randint(0,2,(5,5))
soma = np.sum(array ,axis = 0) #utiliza a formula np.sum para somar e axis=0 somando o valor da coluna

print(f"A soma de cada coluna:{soma}")
```

```python
#7. Crie dois arrays de números aleatórios e encontre a correlação entre eles.
import numpy as np

array_1 = np.random.randint(0,10,10)
array_2 = np.random.randint(0,10,10)

calculo_correlacao = np.corrcoef(array_1, array_2) # correlação entre as duas arrays é uma medida estatistica
extrair_correlacao = calculo_correlacao[0,1]

print(f"Array 1: {array_1}")
print(f"Array 2: {array_2}")
print(f"Correlação entre as arrays: {extrair_correlacao}")
```

```python
#8. Gere uma matriz 3x3 com números inteiros aleatórios e eleve todos os seus elementos ao quadrado.
import numpy as np

array = np.random.randint(1,10,[3,3])

array_exponencial = array * array

print(f"Array: \n{array}")
print(f"Todos elementos ao quadrado: \n{array_exponencial}")
```

```python
#9. Crie um array com 10 números aleatórios e normalize seus valores entre 0 e 1.
import numpy as np

array = np.random.randint(0,100,10)

array_normalizado = (array - np.min(array)) / (np.max(array) - np.min(array)) # para efetuar a normalização (array - np.min(array) / np.max(array) - np.minimo(array))

print(f"Array original: {array}")
print(f"Array normalizado: {array_normalizado}")
```

```python
#10. Crie uma matriz identidade 6x6 e substitua a diagonal principal por valores crescentes de 1 a 6.
import numpy as np

array = np.eye(6) # Gerar uma matriz identidade

np.fill_diagonal(array,[1,2,3,4,5,6]) # Alterar a diagonal principal da array
print(array)
```

```python
#11. Gere um array de 100 números inteiros e encontre o valor que mais se repete no array
import numpy as np

array= np.random.randint(0,101,100)

valores, quantidade = np.unique(array, return_counts= True)

maior_repeticao = valores[np.argmax(quantidade)]

print(array)
print(f"\nO número com maior repetição: {maior_repeticao}")
```

```python
#13. Crie um array contendo os 10 primeiros números pares e calcule sua soma cumulativa.
import numpy as np

array = np.arange(2,21,2) #np.arange(2,21,2) para exibir os 10 primeiros números pares

soma_cumulativa = np.cumsum(array) #np.cumsum para caluclar a soma cumulativa(todas as somas possiveis da array)

print(f"Array: {array}")
print(f"A soma cumulativa: {soma_cumulativa}")
```

```python
#14. Calcule a diferença entre dois arrays de números aleatórios de mesmo comprimento e encontre a média das diferenças.
import numpy as np

array_1 = np.random.randint(0,10,5)
array_2 = np.random.randint(0,10,5)

subtracao = np.subtract(array_1, array_2)
media = np.mean(subtracao)
print(f"Array 1: {array_1} \nArray 2: {array_2}")
print(f"Difença {subtracao}")
print(f"Media: {media}")
```

```python
#16. Crie um array unidimensional e reorganize seus valores em uma matriz 3x4
#array unidemensional é uma array que possui apenas uma dimensão
import numpy as np

array = np.random.randint(0,10,12)


nova_array = np.reshape(array,(3,4)) #np.reshape serve para reogarnizar uma array / dependendo do tamanho da array inical ele não consegue reogarnizar

print(f"Array: {array}")
print(f"Nova array: \n{nova_array}")
```

```python
#17. Gere uma matriz 4x4 de números inteiros aleatórios e multiplique cada valor por sua posição na matriz.
import numpy as np


array = np.random.randint(0,10,(4,4))
posicoes = np.arange(1,17).reshape(4,4) # para gerar a numeração da posição

resultado = array * posicoes # a multiplicação pelo valor * a posicao onde se encontra

print(f"Array: \n{array}")
print(f"Resultado da multplicação: \n{resultado}")
```

```python
#19. Crie dois arrays de números inteiros aleatórios de tamanho 20 e calcule o ângulo entre eles (produto escalar normalizado).
import numpy as np

array_1 = np.random.randint(0,10,20)
array_2 = np.random.randint(0,10,20)

produto_escalar = np.dot(array_1, array_2)

comprimento_vetor_1 = np.linalg.norm(array_1)
comprimento_vetor_2 = np.linalg.norm(array_2)

cosseno_angulo = produto_escalar / (comprimento_vetor_1 * comprimento_vetor_2)

angulo = np.arccos(cosseno_angulo)
angulo_graus = np.degrees(angulo)

print(f"Array 1 : {array_1}")
print(f"Array 2 : {array_2}")
print(f"Produto escalado: {produto_escalar}")
print(f"Ângulo : {angulo_graus} graus")
```
